# 03 — Exploratory Data Analysis (EDA)

Goal: produce decision-relevant insights and define KPIs/segments for the Finance (Lending) problem.


## Core EDA questions (answer these with charts + text)

1. **Portfolio overview**: volume and loan amount distribution by year/month.
2. **Default behavior**: default rate by grade/sub_grade, term, purpose, and issue cohorts.
3. **Pricing vs risk**: how `int_rate` varies across grade/DTI/income.
4. **Borrower profile**: distributions of `annual_inc`, `dti`, `emp_length`; outliers.
5. **Geography**: default rate by state (if `addr_state` exists).
6. **Data quality**: missingness patterns and whether missingness correlates with default.


## KPI definitions (draft)

- **Default rate**: \(\frac{\#\text{defaults}}{\#\text{loans}}\) by segment/time.
- **Avg interest rate**: mean `int_rate` by segment.
- **Avg loan amount**: mean `loan_amnt` by segment.
- **Exposure proxy**: sum of `loan_amnt` by segment (or funded amount if available).
- **Risk segments**: grade/sub_grade + DTI buckets + income buckets.
